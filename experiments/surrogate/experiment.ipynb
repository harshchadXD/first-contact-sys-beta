# First Contact.sys — reproducible surrogate experiment
The saved evaluation is from an actual CPU training run. This notebook is a portable entry point; cells have not been executed inside Jupyter. Decisions in the app use Rust.
Run from a checkout containing frontend dependencies and engine/pkg. Colab/Kaggle CPU is sufficient; NumPy is the only Python dependency. Node must also be installed for label generation.

```python
from pathlib import Path
import subprocess, json
root = Path.cwd()
while not (root / "experiments/surrogate/train.py").exists() and root != root.parent:
    root = root.parent
assert (root / "engine/pkg/first_contact_engine_bg.wasm").exists(), "Open a complete repository checkout"
subprocess.run(["node", "--experimental-strip-types", "experiments/surrogate/generateLabels.mjs"], cwd=root, check=True)
```

```python
import sys
subprocess.run([sys.executable, "experiments/surrogate/train.py"], cwd=root, check=True)
report = json.loads((root / "experiments/surrogate/evaluation.json").read_text())
print(report["deploymentDecision"])
print(json.dumps(report["splits"], indent=2))
```
